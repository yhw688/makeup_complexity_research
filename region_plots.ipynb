```python
import seaborn as sns
```

```python
configs = [
    ("eye_count",         "Eye Region",        "#DD8452"),
    ("lips_count",        "Lips Region",       "#55A868"),
    ("face_count",        "Face Region","#C44E52"),
    ("total_descriptors", "Total Descriptors", "#4C72B0")
]

#2*2
fig, axes = plt.subplots(2, 2, figsize=(12, 10))
axes = axes.flatten()

for ax, (col, title, color) in zip(axes, configs):
    counts = df[col].value_counts().sort_index()
    ax.bar(counts.index, counts.values, color=color, edgecolor="white", linewidth=0.6)
    ax.set_title(f"{title}", fontsize=12, fontweight="bold")
    ax.set_xlabel("Number of descriptors", fontsize=11)
    ax.set_ylabel("Number of images", fontsize=11)
    ax.grid(axis="y", linestyle="--", alpha=0.4)

plt.suptitle("MT-Text Dataset — Descriptor Count Distribution by Region", fontsize=15, fontweight="bold")
plt.tight_layout()

plt.savefig("distribution_plots.png", dpi=150, bbox_inches="tight")
plt.show()
```
